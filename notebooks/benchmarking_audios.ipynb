```python
import matplotlib.pyplot as plt
from pydub import AudioSegment
import librosa
import librosa.display

# Load the audio file using PyDub
audio = AudioSegment.from_file("audio_muestra.mp3")
audio_2 = AudioSegment.from_file("reconstruyendo_audio.mp3")

# Convert the audio to a numpy array
samples = audio.get_array_of_samples()
samples_2 = audio_2.get_array_of_samples()

# Plot the sample graph using Matplotlib
plt.plot(samples)
plt.xlabel('Sample Index')
plt.ylabel('Amplitude')
plt.title('Sample Graph')
plt.show()

plt.plot(samples_2)
plt.xlabel('Sample Index')
plt.ylabel('Amplitude')
plt.title('Sample Graph')
plt.show()
# Alternatively, you can use Librosa to load the audio file and plot the sample graph
# audio, _ = librosa.load("audio_muestra.mp3", sr=None)
# librosa.display.waveplot(audio, sr=None)
# plt.xlabel('Time (s)')
# plt.ylabel('Amplitude')
# plt.title('Sample Graph')
# plt.show()
```
